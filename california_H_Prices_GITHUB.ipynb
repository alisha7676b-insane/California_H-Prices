```python
# --- STEP 3: Handle Categorical Data and Missing Values ---

# Fill missing values in total_bedrooms
california['total_bedrooms'] = california['total_bedrooms'].fillna(
    california['total_bedrooms'].median()
)

# Convert categorical 'ocean_proximity' to numeric columns (One-Hot Encoding)
california_encoded = pd.get_dummies(
    california,
    columns=['ocean_proximity']
)

# Re-define features (X) and target (y) with encoded data
X_encoded = california_encoded.drop('median_house_value', axis=1)
y_encoded = california_encoded['median_house_value']

# Re-split the processed data
X_train_enc, X_test_enc, y_train_enc, y_test_enc = train_test_split(
    X_encoded,
    y_encoded,
    test_size=0.3,
    random_state=42
)


# --- STEP 4: Train Model ---

model = RandomForestRegressor(
    n_estimators=50,
    random_state=42
)

model.fit(X_train_enc, y_train_enc)

print("Model trained successfully!")
```

```python
# ---STEP 4: Evualate---
from sklearn.metrics import mean_squared_error, mean_absolute_error
predictions = model.predict(X_test_enc)


# r^2 Score
r2 = r2_score(y_test_enc, predictions)
print(f"R^2 Score: {r2}")

# Mean Absolute Error (MAE)
mae = mean_absolute_error(y_test_enc, predictions)
print(f"Mean Absolute Error (MAE): {mae:.2f}")

# Mean Sqared Error (MSE)
mae = mean_squared_error(y_test_enc, predictions)
print(f"Mean Squared Error (MSE): {mae:.2f}")
```
